# NexusTrade — Technical Levels Calculator
## Pivot Points & Fibonacci Retracements

Bereken support/resistance levels en Fibonacci retracement zones.

```python
import pandas as pd
import numpy as np

def calculate_pivot_points(high, low, close):
    """
    Calculate Standard Pivot Points
    
    Args:
        high: Period high
        low: Period low
        close: Period close
    
    Returns:
        Dictionary with pivot, support, and resistance levels
    """
    pivot = (high + low + close) / 3
    
    r1 = (2 * pivot) - low
    s1 = (2 * pivot) - high
    
    r2 = pivot + (high - low)
    s2 = pivot - (high - low)
    
    r3 = high + 2 * (pivot - low)
    s3 = low - 2 * (high - pivot)
    
    return {
        'pivot': round(pivot, 4),
        'r1': round(r1, 4),
        'r2': round(r2, 4),
        'r3': round(r3, 4),
        's1': round(s1, 4),
        's2': round(s2, 4),
        's3': round(s3, 4)
    }

def calculate_fibonacci_retracement(high, low, direction='uptrend'):
    """
    Calculate Fibonacci Retracement Levels
    
    Args:
        high: Swing high
        low: Swing low
        direction: 'uptrend' (from low to high) or 'downtrend' (from high to low)
    
    Returns:
        Dictionary with Fibonacci levels
    """
    diff = high - low
    
    if direction == 'uptrend':
        # Retracement from recent high back toward low
        levels = {
            '0%': high,
            '23.6%': high - (0.236 * diff),
            '38.2%': high - (0.382 * diff),
            '50%': high - (0.5 * diff),
            '61.8%': high - (0.618 * diff),
            '78.6%': high - (0.786 * diff),
            '100%': low
        }
    else:  # downtrend
        # Retracement from recent low back toward high
        levels = {
            '0%': low,
            '23.6%': low + (0.236 * diff),
            '38.2%': low + (0.382 * diff),
            '50%': low + (0.5 * diff),
            '61.8%': low + (0.618 * diff),
            '78.6%': low + (0.786 * diff),
            '100%': high
        }
    
    return {k: round(v, 4) for k, v in levels.items()}
```

## Example: DVLT Pivot Points (Daily)

```python
# VOORBEELD: DVLT Daily Pivots (recent session)
high = 0.8375
low = 0.68
close = 0.7386

pivots = calculate_pivot_points(high, low, close)

print("═══════════════════════════════════════════")
print("PIVOT POINTS — DAILY")
print("═══════════════════════════════════════════")
print(f"High:  ${high}")
print(f"Low:   ${low}")
print(f"Close: ${close}")
print("\n" + "─" * 47)
print(f"R3:    ${pivots['r3']}  🚀 Strong resistance")
print(f"R2:    ${pivots['r2']}  ⬆️  Resistance")
print(f"R1:    ${pivots['r1']}  ⬆️  Resistance")
print(f"\nPP:    ${pivots['pivot']}  🎯 Pivot")
print(f"\nS1:    ${pivots['s1']}  ⬇️  Support")
print(f"S2:    ${pivots['s2']}  ⬇️  Support")
print(f"S3:    ${pivots['s3']}  💥 Strong support")
print("═══════════════════════════════════════════")

# Price action context
current = close
if current > pivots['pivot']:
    print(f"\n✅ Price above pivot (${pivots['pivot']}) — bullish bias")
    print(f"   Target R1: ${pivots['r1']}, R2: ${pivots['r2']}")
else:
    print(f"\n🔴 Price below pivot (${pivots['pivot']}) — bearish bias")
    print(f"   Watch S1: ${pivots['s1']}, S2: ${pivots['s2']}")
```

## Example: DVLT Fibonacci Retracement (Uptrend)

```python
# VOORBEELD: DVLT Recent Rally (retracement zones)
swing_low = 0.68    # Recent low (Apr 16)
swing_high = 0.8375  # Recent high (Apr 17-18)

fib_up = calculate_fibonacci_retracement(swing_high, swing_low, 'uptrend')

print("\n═══════════════════════════════════════════")
print("FIBONACCI RETRACEMENT — UPTREND")
print("═══════════════════════════════════════════")
print(f"Swing High: ${swing_high}")
print(f"Swing Low:  ${swing_low}")
print(f"Range:      ${swing_high - swing_low:.4f}")
print("\n" + "─" * 47)
print(f"0%   (High): ${fib_up['0%']:<8}  🚀 Breakout level")
print(f"23.6% Fib:   ${fib_up['23.6%']:<8}  ⬇️  Shallow pullback")
print(f"38.2% Fib:   ${fib_up['38.2%']:<8}  ⬇️  Moderate retracement")
print(f"50.0% Fib:   ${fib_up['50%']:<8}  ⬇️  Half-back level")
print(f"61.8% Fib:   ${fib_up['61.8%']:<8}  💎 Golden Ratio (KEY)")
print(f"78.6% Fib:   ${fib_up['78.6%']:<8}  ⚠️  Deep retracement")
print(f"100% (Low):  ${fib_up['100%']:<8}  💥 Full retracement")
print("═══════════════════════════════════════════")

# Where is current price?
current_price = 0.7386
print(f"\nCurrent Price: ${current_price}")

if current_price >= fib_up['23.6%']:
    print("✅ Above 23.6% — Uptrend intact, shallow pullback")
elif current_price >= fib_up['38.2%']:
    print("⚠️  Between 23.6-38.2% — Moderate pullback zone")
elif current_price >= fib_up['50%']:
    print("⚠️  Between 38.2-50% — Testing support")
elif current_price >= fib_up['61.8%']:
    print("🔴 Between 50-61.8% — Critical support zone (Golden Ratio)")
else:
    print("🔴 Below 61.8% — Uptrend possibly broken")
```

## 52-Week High/Low Analysis

```python
# MU 52-Week Range
week52_high = 1254.81
week52_low = 113.28
current_price = 961.19

fib_52w = calculate_fibonacci_retracement(week52_high, week52_low, 'uptrend')

print("\n═══════════════════════════════════════════")
print("52-WEEK FIBONACCI LEVELS")
print("═══════════════════════════════════════════")
print(f"52W High: ${week52_high}")
print(f"52W Low:  ${week52_low}")
print(f"Current:  ${current_price}")
print("\n" + "─" * 47)

for level, price in fib_52w.items():
    if price == current_price:
        marker = "👉"
    elif abs(price - current_price) < 20:
        marker = "📍"
    else:
        marker = "  "
    print(f"{marker} {level:<6}: ${price:<8.4f}")

print("═══════════════════════════════════════════")

# Calculate position in range
range_pct = ((current_price - week52_low) / (week52_high - week52_low)) * 100
print(f"\nCurrent price is {range_pct:.1f}% of 52-week range")

if range_pct < 25:
    print("🔴 In bottom quartile of 52W range — near lows")
elif range_pct < 50:
    print("⚠️  In lower half of 52W range — below midpoint")
elif range_pct < 75:
    print("✅ In upper half of 52W range — healthy position")
else:
    print("🚀 In top quartile of 52W range — near highs")
```

## Combined Levels: Find Confluence Zones

```python
def find_confluence_zones(pivots, fib_levels, tolerance=0.03):
    """
    Find price zones where pivot and Fibonacci levels align
    (confluence = stronger support/resistance)
    """
    all_levels = []
    
    for name, price in pivots.items():
        all_levels.append(('Pivot ' + name.upper(), price))
    
    for name, price in fib_levels.items():
        all_levels.append(('Fib ' + name, price))
    
    all_levels.sort(key=lambda x: x[1])
    
    confluence = []
    
    for i in range(len(all_levels)):
        for j in range(i + 1, len(all_levels)):
            if abs(all_levels[i][1] - all_levels[j][1]) < tolerance:
                confluence.append({
                    'price': round((all_levels[i][1] + all_levels[j][1]) / 2, 4),
                    'levels': [all_levels[i][0], all_levels[j][0]]
                })
    
    # Remove duplicates
    seen = set()
    unique_confluence = []
    for c in confluence:
        key = c['price']
        if key not in seen:
            seen.add(key)
            unique_confluence.append(c)
    
    return unique_confluence

# Find confluence for MU: daily pivots vs 52-week fib levels (wider tolerance given $800+ price)
confluence = find_confluence_zones(my_pivots, fib_52w, tolerance=15)

print("\n═══════════════════════════════════════════")
print("CONFLUENCE ZONES (High Probability)")
print("═══════════════════════════════════════════")

if confluence:
    for zone in confluence:
        print(f"\n💎 ${zone['price']} — CONFLUENCE ZONE")
        print(f"   Aligning levels:")
        for level in zone['levels']:
            print(f"   - {level}")
        print(f"   ⚠️  Strong support/resistance — expect reaction here")
else:
    print("\nNo strong confluence zones detected within tolerance.")
    print("Pivot and Fib levels are spread out — watch each individually.")

print("\n💡 Key Insight:")
print("Confluence zones = multiple technical levels align at same price.")
print("These act as stronger support/resistance than standalone levels.")
```
