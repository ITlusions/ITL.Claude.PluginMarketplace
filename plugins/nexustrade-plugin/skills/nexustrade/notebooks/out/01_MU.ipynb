# NexusTrade — Position Sizing Calculator
## 1% Risk Rule Calculator

Bereken hoeveel aandelen je moet kopen op basis van portfolio size en risk tolerance.

```python
import pandas as pd
import numpy as np
from datetime import datetime

def calculate_position_size(portfolio_value, risk_percentage, entry_price, stop_loss):
    """
    1% Risk Rule Calculator
    
    Args:
        portfolio_value: Totale portfolio waarde ($)
        risk_percentage: Risk per trade (default 1.0 = 1%)
        entry_price: Entry prijs per share
        stop_loss: Stop loss prijs per share
    
    Returns:
        Dictionary met shares, position value, risk amount
    """
    risk_amount = portfolio_value * (risk_percentage / 100)
    risk_per_share = abs(entry_price - stop_loss)
    
    if risk_per_share == 0:
        return {"error": "Stop loss cannot equal entry price"}
    
    shares = int(risk_amount / risk_per_share)
    position_value = shares * entry_price
    position_pct = (position_value / portfolio_value) * 100
    
    return {
        "shares": shares,
        "position_value": round(position_value, 2),
        "position_pct": round(position_pct, 2),
        "risk_amount": round(risk_amount, 2),
        "risk_per_share": round(risk_per_share, 4)
    }
```

## Example: Single Position Calculation

```python
# MU Trade met 1% risk (illustrative $10,000 portfolio)
portfolio = 10000
entry = 927.50
stop = 897.00
risk_pct = 1.0

result = calculate_position_size(portfolio, risk_pct, entry, stop)

print("═══════════════════════════════════════════")
print("POSITION SIZING — 1% RISK RULE")
print("═══════════════════════════════════════════")
print(f"Portfolio Value: ${portfolio:,.2f}")
print(f"Risk per Trade:  {risk_pct}%")
print(f"Entry Price:     ${entry}")
print(f"Stop Loss:       ${stop}")
print(f"\nRESULTS:")
print(f"Shares to Buy:   {result['shares']}")
print(f"Position Value:  ${result['position_value']:,.2f}")
print(f"Position Size:   {result['position_pct']}% of portfolio")
print(f"Risk Amount:     ${result['risk_amount']}")
print(f"Risk per Share:  ${result['risk_per_share']}")
print("═══════════════════════════════════════════")
```

## Interactive Calculator
Pas de waardes hieronder aan voor jouw specifieke trade:

```python
# ==== PAS DEZE WAARDES AAN (MU trade setup) ====
MY_PORTFOLIO = 10000
MY_RISK_PCT = 1.0
MY_ENTRY = 927.50
MY_STOP = 897.00
# ================================

my_result = calculate_position_size(MY_PORTFOLIO, MY_RISK_PCT, MY_ENTRY, MY_STOP)

print(f"\n🎯 YOUR TRADE SETUP:")
print(f"Buy {my_result['shares']} shares @ ${MY_ENTRY}")
print(f"Stop Loss: ${MY_STOP}")
print(f"Position Size: ${my_result['position_value']} ({my_result['position_pct']}% of portfolio)")
print(f"Max Risk: ${my_result['risk_amount']}\n")

# Warning checks
if my_result['position_pct'] > 10:
    print("⚠️  WARNING: Position >10% is high concentration risk")
if my_result['position_pct'] > 20:
    print("🔴 DANGER: Position >20% violates diversification rules")
if my_result['position_pct'] > 50:
    print("🚨 CRITICAL: Position >50% is gambling, not investing")
```
